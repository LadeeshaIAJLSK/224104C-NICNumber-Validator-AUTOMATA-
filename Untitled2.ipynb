```python

import requests
import pandas as pd

url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": 6.68,
    "longitude": 80.40,
    "start_date": "2010-01-01",
    "end_date": "2025-12-31",
    "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min,precipitation_sum,wind_speed_10m_max",
    "timezone": "Asia/Colombo",
}

r = requests.get(url, params=params, timeout=60)
r.raise_for_status()
df = pd.DataFrame(r.json()["daily"])
df["time"] = pd.to_datetime(df["time"])
df = df.set_index("time")
df.to_csv("ratnapura_daily.csv")
print(df.shape)
df.head()
```

```python
print(df.isna().sum())          # missing values
print(df.describe())            # sanity check ranges
print(df.index.is_monotonic_increasing, df.index.duplicated().sum())
```

```python
monthly = pd.DataFrame({
    "rain_mm": df["precipitation_sum"].resample("MS").sum(),
    "temp_mean": df["temperature_2m_mean"].resample("MS").mean(),
    "temp_max": df["temperature_2m_max"].resample("MS").mean(),
    "temp_min": df["temperature_2m_min"].resample("MS").mean(),
    "wind_max": df["wind_speed_10m_max"].resample("MS").mean(),
})
print(monthly.shape)   # expect 192 rows (16 years x 12)
monthly.head()
```

```python
import matplotlib.pyplot as plt

monthly[["rain_mm", "temp_mean"]].plot(subplots=True, figsize=(12, 6))
plt.show()
```

```python

import numpy as np

data = monthly.copy()

# Lagged values of every variable (only past data, so no leakage)
for col in ["rain_mm", "temp_mean", "wind_max"]:
    for lag in [1, 2, 3, 12]:
        data[f"{col}_lag{lag}"] = data[col].shift(lag)

# Rolling average of the previous 3 months of rain (shift first so the current month is excluded)
data["rain_roll3"] = data["rain_mm"].shift(1).rolling(3).mean()

# Month as a cycle (so December and January are close together)
m = data.index.month
data["month_sin"] = np.sin(2 * np.pi * m / 12)
data["month_cos"] = np.cos(2 * np.pi * m / 12)

# Drop the first 12 rows, which have no lag-12 value
data = data.dropna()
print(data.shape)
```

```python
target = "rain_mm"
features = [c for c in data.columns if "lag" in c or "roll" in c or c.startswith("month_")]

train = data[data.index < "2022-01-01"]
test  = data[data.index >= "2022-01-01"]

X_train, y_train = train[features], train[target]
X_test,  y_test  = test[features],  test[target]

print(len(train), len(test))
print(train.index.max(), test.index.min())
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

def evaluate(y_true, y_pred, name):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"{name:<22} RMSE={rmse:7.1f}  MAE={mae:7.1f}  R²={r2:6.3f}")
    return {"model": name, "RMSE": rmse, "MAE": mae, "R2": r2}

results = []

# Seasonal naive: predict the same month last year
results.append(evaluate(y_test, test["rain_mm_lag12"], "Seasonal naive"))
```

```python

from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor

models = {
    "Random Forest": RandomForestRegressor(
        n_estimators=500, min_samples_leaf=2, random_state=42, n_jobs=-1),
    "XGBoost": XGBRegressor(
        n_estimators=300, learning_rate=0.05, max_depth=3,
        subsample=0.8, colsample_bytree=0.8, random_state=42),
    "LightGBM": LGBMRegressor(
        n_estimators=300, learning_rate=0.05, num_leaves=8,
        min_child_samples=5, random_state=42, verbose=-1),
    "CatBoost": CatBoostRegressor(
        iterations=500, learning_rate=0.05, depth=4,
        random_seed=42, verbose=0),
}

preds = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    preds[name] = model.predict(X_test)
    results.append(evaluate(y_test, preds[name], name))
```

```python
res_df = pd.DataFrame(results).set_index("model").sort_values("RMSE")
display(res_df.round(3))

plt.figure(figsize=(12, 5))
plt.plot(y_test.index, y_test, label="Actual", color="black", linewidth=2)
for name in preds:
    plt.plot(y_test.index, preds[name], label=name, alpha=0.7)
plt.legend()
plt.title("Monthly rainfall: actual vs predicted (test set)")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, name in zip(axes, ["Random Forest", "XGBoost"]):
    imp = pd.Series(models[name].feature_importances_, index=features).sort_values()
    imp.tail(10).plot.barh(ax=ax, title=name)
plt.tight_layout()
plt.show()
```

```python
# SARIMA + hybrid code
from statsmodels.tsa.statespace.sarimax import SARIMAX

y_tr = train["rain_mm"].asfreq("MS")
y_te = test["rain_mm"].asfreq("MS")

sarima = SARIMAX(y_tr, order=(1, 0, 1), seasonal_order=(1, 1, 1, 12)).fit(disp=False)
print(sarima.summary().tables[1])

# One-step-ahead predictions over train + test, using parameters learned on train only
full = pd.concat([y_tr, y_te])
fitted = sarima.apply(full).fittedvalues

sarima_pred = fitted[y_te.index].clip(lower=0)
preds["SARIMA"] = sarima_pred.values
results.append(evaluate(y_te, sarima_pred, "SARIMA"))
```

```python
# Residuals of SARIMA on the TRAINING period only; skip the first 12 months (seasonal differencing warm-up)
resid_train = (y_tr - fitted[y_tr.index]).iloc[12:]

rf_resid = RandomForestRegressor(
    n_estimators=500, min_samples_leaf=3, random_state=42, n_jobs=-1)
rf_resid.fit(X_train.loc[resid_train.index], resid_train)

# Hybrid = SARIMA prediction + RF's predicted correction
hybrid_pred = (fitted[y_te.index] + rf_resid.predict(X_test)).clip(lower=0)
preds["RF + SARIMA"] = hybrid_pred.values
results.append(evaluate(y_te, hybrid_pred, "RF + SARIMA (hybrid)"))
```

```python
res_df = pd.DataFrame(results).set_index("model").sort_values("RMSE")
display(res_df.round(3))

plt.figure(figsize=(12, 5))
plt.plot(y_te.index, y_te, label="Actual", color="black", linewidth=2)
for name in ["Random Forest", "SARIMA", "RF + SARIMA"]:
    plt.plot(y_te.index, preds[name], label=name, alpha=0.8)
plt.legend()
plt.title("Actual vs SARIMA vs RF vs Hybrid (test set)")
plt.show()
```

```python
# ANN + RF/ANN hybrid code
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler().fit(X_train)          # fit on train only
Xtr_s, Xte_s = scaler.transform(X_train), scaler.transform(X_test)

# Average 5 small networks with different seeds, because a single ANN on ~130 rows is unstable
ann_preds = []
for seed in range(5):
    ann = MLPRegressor(hidden_layer_sizes=(16, 8), alpha=1e-2,
                       max_iter=2000, early_stopping=False, random_state=seed)
    ann.fit(Xtr_s, y_train)
    ann_preds.append(ann.predict(Xte_s))

ann_pred = np.clip(np.mean(ann_preds, axis=0), 0, None)
preds["ANN"] = ann_pred
results.append(evaluate(y_test, ann_pred, "ANN"))
```

```python
hybrid_ann = 0.5 * preds["Random Forest"] + 0.5 * ann_pred
preds["RF + ANN"] = hybrid_ann
results.append(evaluate(y_test, hybrid_ann, "RF + ANN (hybrid)"))
```

```python
res_df = pd.DataFrame(results).set_index("model").sort_values("RMSE")
display(res_df.round(3))

res_df["RMSE"].plot.barh(figsize=(8, 5), title="Test RMSE by model (lower is better)")
plt.gca().invert_yaxis()
plt.show()
```

```python
#  stacking ensemble code
from sklearn.base import clone
from sklearn.model_selection import TimeSeriesSplit
from sklearn.linear_model import Ridge

base_names = ["Random Forest", "XGBoost", "LightGBM", "CatBoost"]
tscv = TimeSeriesSplit(n_splits=5)

# Out-of-fold predictions on the TRAINING set only (each fold predicts the future of its own training window)
oof = pd.DataFrame(np.nan, index=X_train.index, columns=base_names)
for tr_idx, va_idx in tscv.split(X_train):
    for name in base_names:
        m = clone(models[name])
        m.fit(X_train.iloc[tr_idx], y_train.iloc[tr_idx])
        oof.iloc[va_idx, oof.columns.get_loc(name)] = m.predict(X_train.iloc[va_idx])

# The first chunk of rows is never predicted out-of-fold, so drop it
mask = oof.notna().all(axis=1)

# Meta-learner: non-negative weights, so it behaves like a sensible weighted average
meta = Ridge(alpha=1.0, positive=True)
meta.fit(oof[mask], y_train[mask])
print("Stack weights:", dict(zip(base_names, meta.coef_.round(3))),
      "| intercept:", round(meta.intercept_, 1))

# Test predictions: base models (already fitted on the full train set in Step 3) -> meta-learner
test_base = np.column_stack([preds[n] for n in base_names])
stack_pred = np.clip(meta.predict(test_base), 0, None)
preds["Stacking"] = stack_pred
results.append(evaluate(y_test, stack_pred, "Stacking (RF+XGB+LGBM+Cat)"))

res_df = pd.DataFrame(results).set_index("model").sort_values("RMSE")
display(res_df.round(3))
```

```python
import pickle, sklearn
print("sklearn version:", sklearn.__version__)   # note this for requirements.txt

best_name = "Random Forest"
pickle.dump({"model": models[best_name], "features": features}, open("model.pkl", "wb"))

monthly.to_csv("monthly.csv")
res_df.to_csv("results.csv")

bt = pd.DataFrame({"actual": y_test})
for n in preds:
    bt[n] = preds[n]
bt.to_csv("backtest.csv")

from google.colab import files
for f in ["model.pkl", "monthly.csv", "results.csv", "backtest.csv"]:
    files.download(f)
```
